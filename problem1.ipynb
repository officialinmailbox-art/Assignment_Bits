```python


# ============================================================
# PROBLEM 1: RNN vs LSTM vs GRU ON SEQUENCES
# Complete coursework implementation
# ============================================================
#
# Google Colab installation:
# !pip install torch numpy pandas matplotlib scikit-learn
#
# Dataset:
# Original UCI HAR Dataset (raw inertial signals)
#
# HAR input: (N, 128, 9)
# Adding input: (N, T, 2)
#
# ============================================================


# ============================================================
# 1. IMPORTS AND CONFIGURATION
# ============================================================

import os
import time
import copy
import random
import zipfile
import urllib.request
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import torch
import torch.nn as nn

from torch.utils.data import Dataset, DataLoader

from sklearn.model_selection import GroupShuffleSplit
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    confusion_matrix,
    ConfusionMatrixDisplay
)

warnings.filterwarnings("ignore")

DEVICE = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Device:", DEVICE)

# Set False for final coursework experiments.
# Set True only for initial debugging.

#QUICK_RUN = True to check that everything executes. Change it to False for final experiments.
QUICK_RUN = True


if QUICK_RUN:
    SEEDS = [42]
    HAR_EPOCHS = 2
    ADD_EPOCHS = 2
    SWEEP_EPOCHS = 2
    ABLATION_EPOCHS = 2
    ADD_TRAIN_SIZE = 1000
    ADD_VAL_SIZE = 250
    ADD_TEST_SIZE = 250
else:
    SEEDS = [11, 22, 33]
    HAR_EPOCHS = 20
    ADD_EPOCHS = 30
    SWEEP_EPOCHS = 15
    ABLATION_EPOCHS = 30
    ADD_TRAIN_SIZE = 6000
    ADD_VAL_SIZE = 1500
    ADD_TEST_SIZE = 1500

BATCH_SIZE = 128
HIDDEN_SIZE = 64
LEARNING_RATE = 1e-3

ADD_LENGTHS = [50, 100, 200, 400]
HIDDEN_SIZES = [16, 32, 64, 128]

STUDENT_ID_LAST_DIGIT = 9

DATA_URL = (
    "https://archive.ics.uci.edu/static/public/240/"
    "human+activity+recognition+using+smartphones.zip"
)

ZIP_FILE = "UCI_HAR_Dataset.zip"
DATA_FOLDER = "UCI HAR Dataset"
RESULTS_DIR = "results"

os.makedirs(RESULTS_DIR, exist_ok=True)


# ============================================================
# 2. REPRODUCIBILITY
# ============================================================

def set_seed(seed):

    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)

    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)

    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False


# ============================================================
# 3. DOWNLOAD AND EXTRACT ORIGINAL UCI HAR
# ============================================================



def download_har_dataset():

    import os
    import zipfile
    import urllib.request

    DATA_URL = (
        "https://archive.ics.uci.edu/static/public/240/"
        "human+activity+recognition+using+smartphones.zip"
    )

    ZIP_FILE = "UCI_HAR_Dataset.zip"
    DATA_FOLDER = "UCI HAR Dataset"

    # Check whether the dataset is already extracted
    required_file = os.path.join(
        DATA_FOLDER,
        "train",
        "Inertial Signals",
        "body_acc_x_train.txt"
    )

    if os.path.isfile(required_file):
        print("UCI HAR dataset already exists.")
        return

    # Remove old download
    if os.path.exists(ZIP_FILE):
        os.remove(ZIP_FILE)

    print("Downloading original UCI HAR dataset...")

    urllib.request.urlretrieve(DATA_URL, ZIP_FILE)

    if not zipfile.is_zipfile(ZIP_FILE):
        raise RuntimeError("Downloaded file is not a valid ZIP.")

    print("Outer ZIP downloaded successfully.")

    # Extract outer ZIP
    with zipfile.ZipFile(ZIP_FILE, "r") as z:
        z.extractall(".")

    print("Outer ZIP extracted.")

    # Check whether the dataset is inside a nested ZIP
    if not os.path.isdir(DATA_FOLDER):

        nested_zip = None

        for root, dirs, files in os.walk("."):

            for filename in files:

                if filename.lower().endswith(".zip"):

                    candidate = os.path.join(root, filename)

                    if candidate == "./" + ZIP_FILE:
                        continue

                    if zipfile.is_zipfile(candidate):
                        nested_zip = candidate
                        break

            if nested_zip:
                break

        if nested_zip is None:
            raise FileNotFoundError(
                "Nested UCI HAR dataset ZIP was not found."
            )

        print("Nested ZIP found:", nested_zip)
        print("Extracting actual HAR dataset...")

        with zipfile.ZipFile(nested_zip, "r") as z:
            z.extractall(".")

    # Final verification
    if not os.path.isfile(required_file):
        raise FileNotFoundError(
            "Dataset extraction failed. Expected file: "
            + required_file
        )

    print("UCI HAR dataset downloaded and extracted successfully!")
    print("Dataset location:", os.path.abspath(DATA_FOLDER))



# ============================================================
# 4. LOAD RAW HAR INERTIAL SIGNALS
# ============================================================

CHANNELS = [
    "body_acc_x",
    "body_acc_y",
    "body_acc_z",
    "body_gyro_x",
    "body_gyro_y",
    "body_gyro_z",
    "total_acc_x",
    "total_acc_y",
    "total_acc_z"
]


def load_signal_split(split):

    base = os.path.join(
        DATA_FOLDER,
        split,
        "Inertial Signals"
    )

    signal_arrays = []

    for channel in CHANNELS:

        filename = f"{channel}_{split}.txt"

        path = os.path.join(
            base,
            filename
        )

        if not os.path.exists(path):
            raise FileNotFoundError(path)

        values = np.loadtxt(path)

        signal_arrays.append(values)

    # Each channel: (N, 128)
    # Stack channels: (N, 128, 9)

    X = np.stack(
        signal_arrays,
        axis=-1
    ).astype(np.float32)

    y_path = os.path.join(
        DATA_FOLDER,
        split,
        f"y_{split}.txt"
    )

    subject_path = os.path.join(
        DATA_FOLDER,
        split,
        f"subject_{split}.txt"
    )

    y = np.loadtxt(
        y_path,
        dtype=int
    ) - 1

    subjects = np.loadtxt(
        subject_path,
        dtype=int
    )

    return X, y, subjects


def load_har_dataset():

    download_har_dataset()

    X_train_all, y_train_all, subject_train = (
        load_signal_split("train")
    )

    X_test, y_test, subject_test = (
        load_signal_split("test")
    )

    print("\nOriginal HAR data loaded.")

    print("Training shape:", X_train_all.shape)
    print("Test shape:", X_test.shape)

    print("Training labels:", np.unique(y_train_all))
    print("Test labels:", np.unique(y_test))

    assert X_train_all.shape[1:] == (128, 9)
    assert X_test.shape[1:] == (128, 9)

    assert len(X_train_all) == len(y_train_all)
    assert len(X_test) == len(y_test)

    assert set(subject_train).isdisjoint(
        set(subject_test)
    )

    return (
        X_train_all,
        y_train_all,
        subject_train,
        X_test,
        y_test,
        subject_test
    )


# ============================================================
# 5. SUBJECT-WISE VALIDATION SPLIT AND NORMALISATION
# ============================================================

class HARDataset(Dataset):

    def __init__(self, X, y):

        self.X = torch.tensor(
            X,
            dtype=torch.float32
        )

        self.y = torch.tensor(
            y,
            dtype=torch.long
        )

    def __len__(self):
        return len(self.y)

    def __getitem__(self, index):
        return self.X[index], self.y[index]


def prepare_har_loaders():

    (
        X_train_all,
        y_train_all,
        subject_train,
        X_test,
        y_test,
        subject_test
    ) = load_har_dataset()

    # Validation subjects are selected only from official
    # training subjects.

    splitter = GroupShuffleSplit(
        n_splits=1,
        test_size=0.20,
        random_state=42
    )

    train_idx, val_idx = next(
        splitter.split(
            X_train_all,
            y_train_all,
            groups=subject_train
        )
    )

    X_train = X_train_all[train_idx]
    y_train = y_train_all[train_idx]
    train_subjects = subject_train[train_idx]

    X_val = X_train_all[val_idx]
    y_val = y_train_all[val_idx]
    val_subjects = subject_train[val_idx]

    # Verify no subject leakage.

    assert set(train_subjects).isdisjoint(
        set(val_subjects)
    )

    assert set(train_subjects).isdisjoint(
        set(subject_test)
    )

    assert set(val_subjects).isdisjoint(
        set(subject_test)
    )

    # Per-channel statistics from TRAINING DATA ONLY.
    # Mean/std are calculated over samples and time steps.

    channel_mean = X_train.mean(
        axis=(0, 1),
        keepdims=True
    )

    channel_std = X_train.std(
        axis=(0, 1),
        keepdims=True
    )

    channel_std[channel_std < 1e-8] = 1.0

    X_train = (
        X_train - channel_mean
    ) / channel_std

    X_val = (
        X_val - channel_mean
    ) / channel_std

    X_test = (
        X_test - channel_mean
    ) / channel_std

    print("\nHAR split sizes:")
    print("Train:", len(X_train))
    print("Validation:", len(X_val))
    print("Test:", len(X_test))

    print("\nTrain subjects:", sorted(set(train_subjects)))
    print("Validation subjects:", sorted(set(val_subjects)))
    print("Test subjects:", sorted(set(subject_test)))

    print("\nPer-channel training means:")
    print(channel_mean.reshape(-1))

    print("\nPer-channel training standard deviations:")
    print(channel_std.reshape(-1))

    loaders = {

        "train": DataLoader(
            HARDataset(X_train, y_train),
            batch_size=BATCH_SIZE,
            shuffle=True
        ),

        "val": DataLoader(
            HARDataset(X_val, y_val),
            batch_size=BATCH_SIZE,
            shuffle=False
        ),

        "test": DataLoader(
            HARDataset(X_test, y_test),
            batch_size=BATCH_SIZE,
            shuffle=False
        )
    }

    metadata = {
        "X_test": X_test,
        "y_test": y_test,
        "channel_mean": channel_mean,
        "channel_std": channel_std
    }

    # Verify exact input shape.

    xb, yb = next(iter(loaders["train"]))

    print("\nHAR batch shape:", xb.shape)
    print("HAR label shape:", yb.shape)

    assert xb.shape[1:] == (128, 9)

    return loaders, metadata


# ============================================================
# 6. HAR CLASS BALANCE PLOT
# ============================================================

ACTIVITY_NAMES = [
    "Walking",
    "Walking upstairs",
    "Walking downstairs",
    "Sitting",
    "Standing",
    "Laying"
]


def plot_class_balance(loaders):

    labels = []

    for _, y in loaders["train"]:
        labels.extend(y.numpy())

    counts = np.bincount(
        labels,
        minlength=6
    )

    plt.figure(figsize=(9, 5))

    plt.bar(
        ACTIVITY_NAMES,
        counts
    )

    plt.xlabel("Activity")
    plt.ylabel("Number of training samples")
    plt.title("HAR Training Class Distribution")

    plt.xticks(rotation=25)
    plt.tight_layout()

    plt.savefig(
        os.path.join(
            RESULTS_DIR,
            "har_class_balance.png"
        )
    )

    plt.show()


# ============================================================
# 7. ADDING PROBLEM GENERATOR
# ============================================================

def generate_adding_data(
    n_samples,
    sequence_length,
    seed
):

    if sequence_length < 2:
        raise ValueError(
            "Sequence length must be at least 2."
        )

    rng = np.random.default_rng(seed)

    values = rng.uniform(
        0,
        1,
        size=(n_samples, sequence_length)
    ).astype(np.float32)

    markers = np.zeros(
        (n_samples, sequence_length),
        dtype=np.float32
    )

    half = sequence_length // 2

    for i in range(n_samples):

        # Exactly one marker in first half.
        first_position = rng.integers(
            0,
            half
        )

        # Exactly one marker in second half.
        second_position = rng.integers(
            half,
            sequence_length
        )

        markers[i, first_position] = 1.0
        markers[i, second_position] = 1.0

    X = np.stack(
        [values, markers],
        axis=-1
    )

    y = (
        values * markers
    ).sum(
        axis=1,
        keepdims=True
    ).astype(np.float32)

    return X, y


# ============================================================
# 8. TEST ADDING DATA GENERATOR
# ============================================================

def test_adding_generator():

    X, y = generate_adding_data(
        n_samples=1000,
        sequence_length=100,
        seed=123
    )

    markers = X[:, :, 1]
    values = X[:, :, 0]

    # Exactly two markers.
    assert np.all(
        markers.sum(axis=1) == 2
    )

    half = X.shape[1] // 2

    # One marker in each half.
    assert np.all(
        markers[:, :half].sum(axis=1) == 1
    )

    assert np.all(
        markers[:, half:].sum(axis=1) == 1
    )

    # Target equals sum of marked values.
    expected = (
        values * markers
    ).sum(
        axis=1,
        keepdims=True
    )

    assert np.allclose(
        y,
        expected,
        atol=1e-6
    )

    print("\nAdding generator tests PASSED.")
    print("Every sequence has exactly two markers.")
    print("One marker is in each half.")
    print("Targets equal the sum of marked values.")


# ============================================================
# 9. ADDING DATA LOADERS
# ============================================================

class SequenceDataset(Dataset):

    def __init__(self, X, y):

        self.X = torch.tensor(
            X,
            dtype=torch.float32
        )

        self.y = torch.tensor(
            y,
            dtype=torch.float32
        )

    def __len__(self):
        return len(self.y)

    def __getitem__(self, index):
        return self.X[index], self.y[index]


def make_adding_loaders(
    sequence_length,
    seed
):

    X_train, y_train = generate_adding_data(
        ADD_TRAIN_SIZE,
        sequence_length,
        seed
    )

    X_val, y_val = generate_adding_data(
        ADD_VAL_SIZE,
        sequence_length,
        seed + 1000
    )

    X_test, y_test = generate_adding_data(
        ADD_TEST_SIZE,
        sequence_length,
        seed + 2000
    )

    return {

        "train": DataLoader(
            SequenceDataset(X_train, y_train),
            batch_size=BATCH_SIZE,
            shuffle=True
        ),

        "val": DataLoader(
            SequenceDataset(X_val, y_val),
            batch_size=BATCH_SIZE,
            shuffle=False
        ),

        "test": DataLoader(
            SequenceDataset(X_test, y_test),
            batch_size=BATCH_SIZE,
            shuffle=False
        )
    }


# ============================================================
# 10. CUSTOM VANILLA RNN CELL
# ============================================================

class MyRNNCell(nn.Module):

    def __init__(self, input_size, hidden_size):

        super().__init__()

        self.weight_ih = nn.Parameter(
            torch.empty(hidden_size, input_size)
        )

        self.weight_hh = nn.Parameter(
            torch.empty(hidden_size, hidden_size)
        )

        self.bias_ih = nn.Parameter(
            torch.zeros(hidden_size)
        )

        self.bias_hh = nn.Parameter(
            torch.zeros(hidden_size)
        )

        nn.init.xavier_uniform_(self.weight_ih)
        nn.init.orthogonal_(self.weight_hh)

    def forward(self, x, h):

        return torch.tanh(
            x @ self.weight_ih.T
            + self.bias_ih
            + h @ self.weight_hh.T
            + self.bias_hh
        )


# ============================================================
# 11. CUSTOM LSTM CELL
# ============================================================

class MyLSTMCell(nn.Module):

    def __init__(
        self,
        input_size,
        hidden_size,
        forget_bias=1.0
    ):

        super().__init__()

        self.hidden_size = hidden_size

        self.weight_ih = nn.Parameter(
            torch.empty(4 * hidden_size, input_size)
        )

        self.weight_hh = nn.Parameter(
            torch.empty(4 * hidden_size, hidden_size)
        )

        self.bias_ih = nn.Parameter(
            torch.zeros(4 * hidden_size)
        )

        self.bias_hh = nn.Parameter(
            torch.zeros(4 * hidden_size)
        )

        nn.init.xavier_uniform_(self.weight_ih)
        nn.init.orthogonal_(self.weight_hh)

        with torch.no_grad():

            # Gate order: input, forget, cell, output.
            self.bias_ih[
                hidden_size:2 * hidden_size
            ].fill_(forget_bias)

    def forward(self, x, state):

        h, c = state

        gates = (
            x @ self.weight_ih.T
            + self.bias_ih
            + h @ self.weight_hh.T
            + self.bias_hh
        )

        i, f, g, o = gates.chunk(4, dim=1)

        i = torch.sigmoid(i)
        f = torch.sigmoid(f)
        g = torch.tanh(g)
        o = torch.sigmoid(o)

        c_new = f * c + i * g

        h_new = o * torch.tanh(c_new)

        return h_new, c_new


# ============================================================
# 12. CUSTOM GRU CELL
# ============================================================

class MyGRUCell(nn.Module):

    def __init__(self, input_size, hidden_size):

        super().__init__()

        self.hidden_size = hidden_size

        self.weight_ih = nn.Parameter(
            torch.empty(3 * hidden_size, input_size)
        )

        self.weight_hh = nn.Parameter(
            torch.empty(3 * hidden_size, hidden_size)
        )

        self.bias_ih = nn.Parameter(
            torch.zeros(3 * hidden_size)
        )

        self.bias_hh = nn.Parameter(
            torch.zeros(3 * hidden_size)
        )

        nn.init.xavier_uniform_(self.weight_ih)
        nn.init.orthogonal_(self.weight_hh)

    def forward(self, x, h):

        input_gates = (
            x @ self.weight_ih.T
            + self.bias_ih
        )

        hidden_gates = (
            h @ self.weight_hh.T
            + self.bias_hh
        )

        i_r, i_z, i_n = input_gates.chunk(3, dim=1)

        h_r, h_z, h_n = hidden_gates.chunk(3, dim=1)

        r = torch.sigmoid(i_r + h_r)

        z = torch.sigmoid(i_z + h_z)

        n = torch.tanh(
            i_n + r * h_n
        )

        h_new = (
            (1 - z) * n
            + z * h
        )

        return h_new


# ============================================================
# 13. CUSTOM CELL SEQUENCE MODEL
# ============================================================

class RecurrentModel(nn.Module):

    def __init__(
        self,
        cell_type,
        input_size,
        hidden_size,
        output_size,
        forget_bias=1.0
    ):

        super().__init__()

        self.cell_type = cell_type
        self.hidden_size = hidden_size

        if cell_type == "RNN":

            self.cell = MyRNNCell(
                input_size,
                hidden_size
            )

        elif cell_type == "LSTM":

            self.cell = MyLSTMCell(
                input_size,
                hidden_size,
                forget_bias
            )

        elif cell_type == "GRU":

            self.cell = MyGRUCell(
                input_size,
                hidden_size
            )

        else:
            raise ValueError("Unknown cell type.")

        self.fc = nn.Linear(
            hidden_size,
            output_size
        )

    def forward(
        self,
        x,
        return_states=False
    ):

        batch_size = x.size(0)

        h = torch.zeros(
            batch_size,
            self.hidden_size,
            device=x.device
        )

        if self.cell_type == "LSTM":
            c = torch.zeros_like(h)

        states = []

        for t in range(x.size(1)):

            x_t = x[:, t, :]

            if self.cell_type == "LSTM":

                h, c = self.cell(
                    x_t,
                    (h, c)
                )

            else:

                h = self.cell(
                    x_t,
                    h
                )

            if return_states:
                states.append(h)

        output = self.fc(h)

        if return_states:
            return output, states

        return output


def count_parameters(model):

    return sum(
        p.numel()
        for p in model.parameters()
        if p.requires_grad
    )


# ============================================================
# 14. CELL EQUIVALENCE TESTS
# ============================================================

def copy_cell_weights(custom, library):

    """
    Copy PyTorch library weights into custom recurrent cells.
    """

    with torch.no_grad():

        custom.weight_ih.copy_(
            library.weight_ih_l0
        )

        custom.weight_hh.copy_(
            library.weight_hh_l0
        )

        custom.bias_ih.copy_(
            library.bias_ih_l0
        )

        custom.bias_hh.copy_(
            library.bias_hh_l0
        )


def test_cell_equivalence():

    torch.manual_seed(42)

    input_size = 9
    hidden_size = 64
    batch_size = 8
    seq_len = 10

    x = torch.randn(
        batch_size,
        seq_len,
        input_size
    )

    print("\n========== CELL EQUIVALENCE TESTS ==========\n")

    # ========================================================
    # 1. RNN TEST
    # ========================================================

    custom_rnn = MyRNNCell(
        input_size,
        hidden_size
    )

    torch_rnn = nn.RNN(
        input_size=input_size,
        hidden_size=hidden_size,
        nonlinearity="tanh",
        batch_first=True
    )

    copy_cell_weights(
        custom_rnn,
        torch_rnn
    )

    h0 = torch.randn(
        batch_size,
        hidden_size
    )

    h_custom = h0

    for t in range(seq_len):

        h_custom = custom_rnn(
            x[:, t, :],
            h_custom
        )

    _, h_torch = torch_rnn(
        x,
        h0.unsqueeze(0)
    )

    rnn_diff = (
        h_custom - h_torch.squeeze(0)
    ).abs().max().item()

    print("RNN maximum difference:", rnn_diff)

    assert rnn_diff < 1e-5, (
        "RNN equivalence test failed!"
    )

    # ========================================================
    # 2. LSTM TEST
    # ========================================================

    custom_lstm = MyLSTMCell(
        input_size,
        hidden_size
    )

    torch_lstm = nn.LSTM(
        input_size=input_size,
        hidden_size=hidden_size,
        batch_first=True
    )

    copy_cell_weights(
        custom_lstm,
        torch_lstm
    )

    h0_lstm = torch.randn(
        batch_size,
        hidden_size
    )

    c0_lstm = torch.randn(
        batch_size,
        hidden_size
    )

    h_custom = h0_lstm
    c_custom = c0_lstm

    for t in range(seq_len):

        h_custom, c_custom = custom_lstm(
            x[:, t, :],
            (h_custom, c_custom)
        )

    _, (h_torch, c_torch) = torch_lstm(
        x,
        (
            h0_lstm.unsqueeze(0),
            c0_lstm.unsqueeze(0)
        )
    )

    lstm_h_diff = (
        h_custom - h_torch.squeeze(0)
    ).abs().max().item()

    lstm_c_diff = (
        c_custom - c_torch.squeeze(0)
    ).abs().max().item()

    print("LSTM hidden-state difference:", lstm_h_diff)
    print("LSTM cell-state difference:", lstm_c_diff)

    assert lstm_h_diff < 1e-5, (
        "LSTM hidden-state test failed!"
    )

    assert lstm_c_diff < 1e-5, (
        "LSTM cell-state test failed!"
    )

    # ========================================================
    # 3. GRU TEST
    # ========================================================

    custom_gru = MyGRUCell(
        input_size,
        hidden_size
    )

    torch_gru = nn.GRU(
        input_size=input_size,
        hidden_size=hidden_size,
        batch_first=True
    )

    copy_cell_weights(
        custom_gru,
        torch_gru
    )

    h0_gru = torch.randn(
        batch_size,
        hidden_size
    )

    h_custom = h0_gru

    for t in range(seq_len):

        h_custom = custom_gru(
            x[:, t, :],
            h_custom
        )

    _, h_torch = torch_gru(
        x,
        h0_gru.unsqueeze(0)
    )

    gru_diff = (
        h_custom - h_torch.squeeze(0)
    ).abs().max().item()

    print("GRU maximum difference:", gru_diff)

    assert gru_diff < 1e-5, (
        "GRU equivalence test failed!"
    )

    print("\nAll three cell equivalence tests PASSED!")



# ============================================================
# 15. PARAMETER COUNT DERIVATION
# ============================================================

def parameter_count_comparison():

    d = 9
    h = 64

    # Hand counts use one bias vector per gate.
    manual = {

        "RNN": h * d + h * h + h,

        "LSTM": (
            4 * h * d
            + 4 * h * h
            + 4 * h
        ),

        "GRU": (
            3 * h * d
            + 3 * h * h
            + 3 * h
        )
    }

    # PyTorch stores two bias vectors.
    pytorch = {

        "RNN": h * d + h * h + 2 * h,

        "LSTM": (
            4 * h * d
            + 4 * h * h
            + 8 * h
        ),

        "GRU": (
            3 * h * d
            + 3 * h * h
            + 6 * h
        )
    }

    print("\nParameter count comparison")
    print("d =", d, "| h =", h)

    for name in ["RNN", "LSTM", "GRU"]:

        print(
            f"{name}: "
            f"Manual={manual[name]}, "
            f"PyTorch={pytorch[name]}"
        )

    assert manual["RNN"] == 4736
    assert manual["LSTM"] == 18944
    assert manual["GRU"] == 14208

    assert pytorch["RNN"] == 4800
    assert pytorch["LSTM"] == 19200
    assert pytorch["GRU"] == 14400


# ============================================================
# 16. TRAINING AND EVALUATION FUNCTIONS
# ============================================================

def run_epoch(
    model,
    loader,
    task,
    optimizer=None,
    grad_clip=1.0
):

    training = optimizer is not None

    model.train() if training else model.eval()

    criterion = (
        nn.CrossEntropyLoss()
        if task == "classification"
        else nn.MSELoss()
    )

    total_loss = 0.0
    predictions = []
    targets = []

    start = time.time()

    for X, y in loader:

        X = X.to(DEVICE)
        y = y.to(DEVICE)

        if training:
            optimizer.zero_grad()

        with torch.set_grad_enabled(training):

            output = model(X)

            loss = criterion(
                output,
                y
            )

            if training:

                loss.backward()

                if grad_clip is not None:

                    torch.nn.utils.clip_grad_norm_(
                        model.parameters(),
                        max_norm=grad_clip
                    )

                optimizer.step()

        total_loss += (
            loss.item() * len(y)
        )

        if task == "classification":

            predictions.extend(
                output.argmax(dim=1).detach().cpu().numpy()
            )

            targets.extend(
                y.detach().cpu().numpy()
            )

    elapsed = time.time() - start

    average_loss = (
        total_loss / len(loader.dataset)
    )

    if task == "classification":

        accuracy = accuracy_score(
            targets,
            predictions
        )

        macro_f1 = f1_score(
            targets,
            predictions,
            average="macro",
            zero_division=0
        )

        return {
            "loss": average_loss,
            "accuracy": accuracy,
            "macro_f1": macro_f1,
            "seconds": elapsed,
            "predictions": np.asarray(predictions),
            "targets": np.asarray(targets)
        }

    return {
        "loss": average_loss,
        "seconds": elapsed
    }


# ============================================================
# 17. HAR CLASSIFICATION TRAINING
# ============================================================

def train_har_model(
    cell_type,
    loaders,
    seed,
    hidden_size=HIDDEN_SIZE,
    epochs=HAR_EPOCHS
):

    set_seed(seed)

    model = RecurrentModel(
        cell_type,
        input_size=9,
        hidden_size=hidden_size,
        output_size=6
    ).to(DEVICE)

    optimizer = torch.optim.Adam(
        model.parameters(),
        lr=LEARNING_RATE
    )

    best_val_f1 = -1
    best_state = None

    epoch_times = []

    for epoch in range(epochs):

        train_result = run_epoch(
            model,
            loaders["train"],
            "classification",
            optimizer
        )

        val_result = run_epoch(
            model,
            loaders["val"],
            "classification"
        )

        epoch_times.append(
            train_result["seconds"]
        )

        if val_result["macro_f1"] > best_val_f1:

            best_val_f1 = val_result["macro_f1"]

            best_state = copy.deepcopy(
                model.state_dict()
            )

        print(
            f"{cell_type} | seed={seed} | "
            f"epoch={epoch + 1}/{epochs} | "
            f"train loss={train_result['loss']:.4f} | "
            f"val F1={val_result['macro_f1']:.4f}"
        )

    model.load_state_dict(best_state)

    test_result = run_epoch(
        model,
        loaders["test"],
        "classification"
    )

    return {

        "cell": cell_type,
        "seed": seed,
        "hidden_size": hidden_size,
        "test_accuracy": test_result["accuracy"],
        "test_macro_f1": test_result["macro_f1"],
        "best_val_macro_f1": best_val_f1,
        "parameters": count_parameters(model),
        "seconds_per_epoch": np.mean(epoch_times),
        "predictions": test_result["predictions"],
        "targets": test_result["targets"],
        "state_dict": best_state

    }


# ============================================================
# 18. RUN THREE-SEED HAR EXPERIMENTS
# ============================================================

def run_har_experiments(loaders):

    results = []
    best_by_cell = {}

    for cell in ["RNN", "LSTM", "GRU"]:

        for seed in SEEDS:

            print(
                f"\nHAR: Training {cell}, seed={seed}"
            )

            result = train_har_model(
                cell,
                loaders,
                seed
            )

            results.append({
                k: v for k, v in result.items()
                if k not in [
                    "predictions",
                    "targets",
                    "state_dict"
                ]
            })

            if (
                cell not in best_by_cell
                or result["best_val_macro_f1"]
                > best_by_cell[cell]["best_val_macro_f1"]
            ):

                best_by_cell[cell] = result

    df = pd.DataFrame(results)

    df.to_csv(
        os.path.join(
            RESULTS_DIR,
            "har_all_seed_results.csv"
        ),
        index=False
    )

    summary = df.groupby("cell").agg(
        accuracy_mean=("test_accuracy", "mean"),
        accuracy_std=("test_accuracy", "std"),
        macro_f1_mean=("test_macro_f1", "mean"),
        macro_f1_std=("test_macro_f1", "std"),
        parameters=("parameters", "first"),
        seconds_per_epoch=("seconds_per_epoch", "mean")
    ).reset_index()

    summary.to_csv(
        os.path.join(
            RESULTS_DIR,
            "har_summary.csv"
        ),
        index=False
    )

    print("\nHAR MEAN ± STD")
    print(summary.to_string(index=False))

    # Best model selected by validation macro-F1.
    best_cell = max(
        best_by_cell,
        key=lambda c: best_by_cell[c]["best_val_macro_f1"]
    )

    best_result = best_by_cell[best_cell]

    cm = confusion_matrix(
        best_result["targets"],
        best_result["predictions"],
        labels=list(range(6))
    )

    display = ConfusionMatrixDisplay(
        confusion_matrix=cm,
        display_labels=ACTIVITY_NAMES
    )

    fig, ax = plt.subplots(figsize=(9, 8))

    display.plot(
        ax=ax,
        xticks_rotation=35,
        cmap="Blues",
        colorbar=False
    )

    plt.title(
        f"Best HAR Model: {best_cell}"
    )

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            RESULTS_DIR,
            "har_best_confusion_matrix.png"
        )
    )

    plt.show()

    return df, summary, best_by_cell


# ============================================================
# 19. ADDING TRAINING WITH LEARNING CURVES
# ============================================================

def train_adding_model(
    cell_type,
    loaders,
    seed,
    sequence_length,
    epochs=ADD_EPOCHS,
    hidden_size=HIDDEN_SIZE,
    forget_bias=1.0,
    grad_clip=1.0,
    save_curves=False
):

    set_seed(seed)

    model = RecurrentModel(
        cell_type,
        input_size=2,
        hidden_size=hidden_size,
        output_size=1,
        forget_bias=forget_bias
    ).to(DEVICE)

    optimizer = torch.optim.Adam(
        model.parameters(),
        lr=LEARNING_RATE
    )

    best_val = float("inf")
    best_state = None

    train_curve = []
    val_curve = []

    for epoch in range(epochs):

        train_result = run_epoch(
            model,
            loaders["train"],
            "regression",
            optimizer,
            grad_clip=grad_clip
        )

        val_result = run_epoch(
            model,
            loaders["val"],
            "regression"
        )

        train_curve.append(
            train_result["loss"]
        )

        val_curve.append(
            val_result["loss"]
        )

        if val_result["loss"] < best_val:

            best_val = val_result["loss"]

            best_state = copy.deepcopy(
                model.state_dict()
            )

        if (
            save_curves
            or epoch == 0
            or (epoch + 1) % 5 == 0
        ):

            print(
                f"{cell_type} | T={sequence_length} | "
                f"epoch={epoch + 1}/{epochs} | "
                f"train MSE={train_result['loss']:.5f} | "
                f"val MSE={val_result['loss']:.5f}"
            )

    model.load_state_dict(best_state)

    test_result = run_epoch(
        model,
        loaders["test"],
        "regression"
    )

    result = {

        "cell": cell_type,
        "seed": seed,
        "sequence_length": sequence_length,
        "test_mse": test_result["loss"],
        "best_val_mse": best_val,
        "solved": test_result["loss"] < 0.05,
        "parameters": count_parameters(model)

    }

    if save_curves:

        result["train_curve"] = train_curve
        result["val_curve"] = val_curve

    return result


# ============================================================
# 20. ADDING PROBLEM EXPERIMENTS
# ============================================================

def run_adding_experiments():

    results = []
    curves_t200 = {}

    for T in ADD_LENGTHS:

        for seed in SEEDS:

            loaders = make_adding_loaders(
                T,
                seed
            )

            for cell in ["RNN", "LSTM", "GRU"]:

                print(
                    f"\nAdding: {cell}, T={T}, seed={seed}"
                )

                save_curves = (
                    T == 200 and seed == SEEDS[0]
                )

                result = train_adding_model(
                    cell,
                    loaders,
                    seed,
                    T,
                    save_curves=save_curves
                )

                if save_curves:

                    curves_t200[cell] = result

                results.append({
                    k: v for k, v in result.items()
                    if k not in [
                        "train_curve",
                        "val_curve"
                    ]
                })

    df = pd.DataFrame(results)

    df.to_csv(
        os.path.join(
            RESULTS_DIR,
            "adding_all_results.csv"
        ),
        index=False
    )

    summary = df.groupby(
        ["cell", "sequence_length"]
    ).agg(
        mse_mean=("test_mse", "mean"),
        mse_std=("test_mse", "std"),
        solved_runs=("solved", "sum"),
        total_runs=("solved", "count")
    ).reset_index()

    summary.to_csv(
        os.path.join(
            RESULTS_DIR,
            "adding_summary.csv"
        ),
        index=False
    )

    print("\nADDING PROBLEM MEAN ± STD")
    print(summary.to_string(index=False))

    # Constant optimal prediction baseline:
    # E[target] = 1, and MSE = Var(target) = 1/6.
    baseline = 1.0 / 6.0

    print(
        "\nConstant-prediction baseline MSE:",
        baseline
    )

    # Plot learning curves for T=200.
    plt.figure(figsize=(10, 6))

    for cell, result in curves_t200.items():

        plt.plot(
            result["train_curve"],
            label=f"{cell} train"
        )

        plt.plot(
            result["val_curve"],
            linestyle="--",
            label=f"{cell} validation"
        )

    plt.axhline(
        baseline,
        linestyle=":",
        label="Constant baseline (1/6)"
    )

    plt.xlabel("Epoch")
    plt.ylabel("MSE")

    plt.title(
        "Adding Problem Learning Curves (T=200)"
    )

    plt.legend()
    plt.grid(True)

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            RESULTS_DIR,
            "adding_learning_curves_T200.png"
        )
    )

    plt.show()

    # Solved runs table.
    solved_table = summary.pivot(
        index="sequence_length",
        columns="cell",
        values="solved_runs"
    )

    solved_table.to_csv(
        os.path.join(
            RESULTS_DIR,
            "adding_solved_runs.csv"
        )
    )

    print("\nSOLVED RUNS (MSE < 0.05)")
    print(solved_table)

    return df, summary, curves_t200


# ============================================================
# 21. GRADIENT PROFILE AT T=100
# ============================================================

def get_gradient_profile(
    cell_type,
    trained,
    seed=42,
    sequence_length=100
):

    set_seed(seed)

    loaders = make_adding_loaders(
        sequence_length,
        seed
    )

    model = RecurrentModel(
        cell_type,
        input_size=2,
        hidden_size=HIDDEN_SIZE,
        output_size=1
    ).to(DEVICE)

    if trained:

        optimizer = torch.optim.Adam(
            model.parameters(),
            lr=LEARNING_RATE
        )

        for _ in range(ADD_EPOCHS):

            run_epoch(
                model,
                loaders["train"],
                "regression",
                optimizer,
                grad_clip=1.0
            )

    X, y = next(iter(loaders["test"]))

    X = X.to(DEVICE)
    y = y.to(DEVICE)

    model.zero_grad()

    output, states = model(
        X,
        return_states=True
    )

    for state in states:
        state.retain_grad()

    loss = nn.MSELoss()(
        output,
        y
    )

    loss.backward()

    norms = []

    for state in states:

        if state.grad is None:
            norms.append(0.0)
        else:
            norms.append(
                state.grad.norm().item()
            )

    return np.asarray(norms)


def plot_gradient_profiles():

    for cell in ["RNN", "LSTM", "GRU"]:

        before = get_gradient_profile(
            cell,
            trained=False
        )

        after = get_gradient_profile(
            cell,
            trained=True
        )

        plt.figure(figsize=(9, 5))

        plt.plot(
            np.arange(1, 101),
            np.maximum(before, 1e-12),
            label="Before training"
        )

        plt.plot(
            np.arange(1, 101),
            np.maximum(after, 1e-12),
            label="After training"
        )

        plt.yscale("log")

        plt.xlabel("Time step t")
        plt.ylabel("||dL / dh_t|| (log scale)")

        plt.title(
            f"{cell}: Hidden-state gradient at T=100"
        )

        plt.legend()
        plt.grid(True, which="both")

        plt.tight_layout()

        plt.savefig(
            os.path.join(
                RESULTS_DIR,
                f"gradient_T100_{cell}.png"
            )
        )

        plt.show()


# ============================================================
# 22. HIDDEN-SIZE SWEEP ON HAR
# ============================================================

def run_hidden_size_sweep(loaders):

    results = []

    for cell in ["LSTM", "GRU"]:

        for hidden_size in HIDDEN_SIZES:

            for seed in SEEDS:

                print(
                    f"\nSweep: {cell}, hidden={hidden_size}, "
                    f"seed={seed}"
                )

                result = train_har_model(
                    cell,
                    loaders,
                    seed,
                    hidden_size=hidden_size,
                    epochs=SWEEP_EPOCHS
                )

                results.append({
                    k: v for k, v in result.items()
                    if k not in [
                        "predictions",
                        "targets",
                        "state_dict"
                    ]
                })

    df = pd.DataFrame(results)

    df.to_csv(
        os.path.join(
            RESULTS_DIR,
            "har_hidden_size_sweep.csv"
        ),
        index=False
    )

    summary = df.groupby(
        ["cell", "hidden_size"]
    ).agg(
        accuracy_mean=("test_accuracy", "mean"),
        accuracy_std=("test_accuracy", "std"),
        f1_mean=("test_macro_f1", "mean"),
        f1_std=("test_macro_f1", "std"),
        parameters=("parameters", "first"),
        seconds_per_epoch=("seconds_per_epoch", "mean")
    ).reset_index()

    print("\nHIDDEN SIZE SWEEP SUMMARY")
    print(summary.to_string(index=False))

    summary.to_csv(
        os.path.join(
            RESULTS_DIR,
            "har_hidden_size_summary.csv"
        ),
        index=False
    )

    plt.figure(figsize=(9, 5))

    for cell in ["LSTM", "GRU"]:

        subset = summary[
            summary["cell"] == cell
        ]

        plt.plot(
            subset["hidden_size"],
            subset["accuracy_mean"],
            marker="o",
            label=cell
        )

    plt.xlabel("Hidden size")
    plt.ylabel("Mean test accuracy")

    plt.title("HAR Hidden-Size Sweep")

    plt.legend()
    plt.grid(True)

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            RESULTS_DIR,
            "har_hidden_size_sweep.png"
        )
    )

    plt.show()

    return summary


# ============================================================
# 23. ODD STUDENT-ID ABLATION:
# LSTM FORGET BIAS 0 VS 1 AT T=200
# ============================================================

def run_individual_ablation():

    if STUDENT_ID_LAST_DIGIT % 2 == 0:

        print(
            "Even student ID: assignment requires "
            "RNN gradient-clipping ablation."
        )

        return None

    print(
        "\nOdd student ID: LSTM forget-bias ablation."
    )

    T = 200

    results = []

    for forget_bias in [0.0, 1.0]:

        for seed in SEEDS:

            loaders = make_adding_loaders(
                T,
                seed
            )

            set_seed(seed)

            model = RecurrentModel(
                "LSTM",
                input_size=2,
                hidden_size=HIDDEN_SIZE,
                output_size=1,
                forget_bias=forget_bias
            ).to(DEVICE)

            optimizer = torch.optim.Adam(
                model.parameters(),
                lr=LEARNING_RATE
            )

            best_val = float("inf")
            best_state = None

            for epoch in range(ABLATION_EPOCHS):

                train_result = run_epoch(
                    model,
                    loaders["train"],
                    "regression",
                    optimizer,
                    grad_clip=1.0
                )

                val_result = run_epoch(
                    model,
                    loaders["val"],
                    "regression"
                )

                if val_result["loss"] < best_val:

                    best_val = val_result["loss"]

                    best_state = copy.deepcopy(
                        model.state_dict()
                    )

                print(
                    f"Forget bias={forget_bias} | "
                    f"seed={seed} | "
                    f"epoch={epoch + 1} | "
                    f"val MSE={val_result['loss']:.5f}"
                )

            model.load_state_dict(best_state)

            test_result = run_epoch(
                model,
                loaders["test"],
                "regression"
            )

            results.append({

                "forget_bias": forget_bias,
                "seed": seed,
                "sequence_length": T,
                "test_mse": test_result["loss"],
                "best_val_mse": best_val,
                "solved": test_result["loss"] < 0.05

            })

    df = pd.DataFrame(results)

    df.to_csv(
        os.path.join(
            RESULTS_DIR,
            "lstm_forget_bias_ablation.csv"
        ),
        index=False
    )

    summary = df.groupby(
        "forget_bias"
    ).agg(
        mse_mean=("test_mse", "mean"),
        mse_std=("test_mse", "std"),
        solved_runs=("solved", "sum")
    ).reset_index()

    print("\nLSTM FORGET-BIAS ABLATION")
    print(summary.to_string(index=False))

    return df, summary


# ============================================================
# 24. GENERATE ANALYSIS REPORT TEMPLATE
# ============================================================

def save_analysis_template(
    har_summary,
    adding_summary,
    sweep_summary,
    ablation_result
):

    path = os.path.join(
        RESULTS_DIR,
        "analysis_report_template.txt"
    )

    with open(path, "w") as f:

        f.write(
            "PROBLEM 1: RNN vs LSTM vs GRU ANALYSIS\n\n"
        )

        f.write(
            "1. HAR CLASSIFICATION\n"
        )

        f.write(
            har_summary.to_string(index=False)
        )

        f.write("\n\n2. ADDING PROBLEM\n")

        f.write(
            adding_summary.to_string(index=False)
        )

        f.write(
            "\n\n3. HIDDEN SIZE SWEEP\n"
        )

        f.write(
            sweep_summary.to_string(index=False)
        )

        f.write(
            "\n\n4. INDIVIDUAL ABLATION\n"
        )

        if ablation_result is not None:

            f.write(
                str(ablation_result)
            )

        f.write(
            "\n\n5. DISCUSSION POINTS TO COMPLETE\n"
        )

        f.write(
            """
A. Gradient and memory:
Discuss the gradient plots at T=100.
Explain whether the RNN gradients vanish as time steps increase.
Compare the gradient preservation of LSTM and GRU.

B. Performance as T grows:
Use the adding-problem MSE table.
Identify which models solve T=50, 100, 200 and 400.
A run is solved only when test MSE < 0.05.

C. HAR accuracy, parameters and time:
Compare mean test accuracy and macro-F1.
Compare parameter counts and seconds per epoch.
Explain the accuracy/compute trade-off.

D. Wearable recommendation:
Use your measured accuracy, parameter count and runtime.
Discuss battery and memory constraints.
Do not claim a model is best without referring to your results.

E. Ablation:
For odd student ID, compare LSTM forget bias 0 and 1.
Explain how forget-gate initialisation changes the initial
tendency to retain or forget cell-state information.
"""
        )

    print(
        "\nAnalysis template saved:",
        path
    )


# ============================================================
# 25. MAIN PROGRAM
# ============================================================

def main():

    print("=" * 70)
    print("PROBLEM 1: RNN vs LSTM vs GRU")
    print("=" * 70)

    print("Device:", DEVICE)
    print("Seeds:", SEEDS)
    print("Quick run:", QUICK_RUN)

    # A. Verify custom cells and parameter counts.

    print("\nA. CELL EQUIVALENCE")

    test_cell_equivalence()

    print("\nB. PARAMETER COUNTS")

    parameter_count_comparison()

    # C. Adding generator correctness.

    print("\nC. ADDING GENERATOR TEST")

    test_adding_generator()

    # D. HAR pipeline.

    print("\nD. HAR DATA PIPELINE")

    har_loaders, har_metadata = prepare_har_loaders()

    plot_class_balance(har_loaders)

    # E. HAR experiments.

    print("\nE. HAR CLASSIFICATION")

    har_results, har_summary, best_models = (
        run_har_experiments(har_loaders)
    )

    # F. Adding experiments.

    print("\nF. ADDING PROBLEM")

    adding_results, adding_summary, adding_curves = (
        run_adding_experiments()
    )

    # G. Gradient analysis.

    print("\nG. GRADIENT ANALYSIS AT T=100")

    plot_gradient_profiles()

    # H. Hidden-size sweep.

    print("\nH. HIDDEN-SIZE SWEEP")

    sweep_summary = run_hidden_size_sweep(
        har_loaders
    )

    # I. Individualisation.

    print("\nI. INDIVIDUAL ABLATION")

    ablation_result = run_individual_ablation()

    # J. Analysis template.

    save_analysis_template(
        har_summary,
        adding_summary,
        sweep_summary,
        ablation_result
    )

    # K. List all output files.

    print("\n" + "=" * 70)
    print("ALL EXPERIMENTS COMPLETED")
    print("=" * 70)

    print("\nResults folder:", RESULTS_DIR)

    for filename in sorted(
        os.listdir(RESULTS_DIR)
    ):

        print("-", filename)


# ============================================================
# 26. RUN
# ============================================================

if __name__ == "__main__":
    main()
```
